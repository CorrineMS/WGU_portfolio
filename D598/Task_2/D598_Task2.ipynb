```python
# Import pandas and load the dataset
import pandas as pd

df = pd.read_excel("D598 Data Set.xlsx")
df.head()
```

```python
# Identify duplicate rows in the dataset
duplicates = df[df.duplicated()]
duplicates
```

```python
# Group data by Business State and calculate descriptive statistics
grouped_stats = (
    df.groupby("Business State")
      .agg(["mean", "median", "min", "max"])
)

grouped_stats
```

```python
# Filter businesses with negative debt-to-equity ratios
negative_debt_to_equity = df[df["Debt to Equity"] < 0]

negative_debt_to_equity
```

```python
import numpy as np

# Create a new dataframe for debt-to-income ratio
debt_to_income_df = df[["Total Long-term Debt", "Total Revenue"]].copy()

# Prevent division by zero using explicit conditional handling
debt_to_income_df["Debt-to-Income Ratio"] = np.where(
    debt_to_income_df["Total Revenue"] == 0,
    np.nan,
    debt_to_income_df["Total Long-term Debt"] / debt_to_income_df["Total Revenue"]
)

debt_to_income_df
```

```python
# Concatenate the debt-to-income ratio dataframe with the original dataframe
df_final = pd.concat(
    [df, debt_to_income_df[["Debt-to-Income Ratio"]]],
    axis=1
)

df_final
```
